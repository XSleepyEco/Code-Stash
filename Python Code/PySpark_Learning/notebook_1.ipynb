```python
## read the dataset option 2
df_pyspark = spark.read \
    .option("header", "true") \
    .csv("Water Quality - Every Drop Matters\Cities1.csv", inferSchema=True)

df_pyspark.printSchema() 
```

inferSchema tries to automatically determine the data types of the fields.

```python
## Adding columns in pyspark dataframes
df_pyspark.withColumn('AirQuality divided', df_pyspark['AirQuality'] / 10)
```

```python
df_pyspark.withColumn('AirQuality divided', df_pyspark['AirQuality'] / 10).show()
```

```python
df_pyspark = df_pyspark.withColumn('AirQuality divided', df_pyspark['AirQuality'] / 10)
```

```python
df_pyspark.show()
```

```python
## dropping columns
df_pyspark.drop('AirQuality divided')
```

```python
df_pyspark = df_pyspark.drop('AirQuality divided')
```

```python
df_pyspark.show()
```

```python
## renaming columns 
df_pyspark.withColumnRenamed('City', 'RenamedCity')
```
