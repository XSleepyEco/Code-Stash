```python
#how="any", removes any row where at least one attribute has a null, any is the default value
df_pyspark.na.drop(how="any").show() 
```

```python
#how=="all", removes any row where all the attributes are null
df_pyspark.na.drop(how="all").show() 
```

```python
#Threshold - at least n non-null value must be present otherwise drop the row.
n = 5
df_pyspark.na.drop(thresh=n).show()
```

```python
#Subset - if there is any nulls in a column (a list of column names) drop the row
df_pyspark.na.drop(subset=['Region']).show()
```

```python
#filling the missing values - wherever theere is an null value than replace it with another value
df_pyspark.na.fill('Missing Values').show()
```

```python
# a list can be given to specify which columns to replace null values from
df_pyspark.na.fill('Missing Values', ['City','Country']).show()
```

```python
df_pyspark.show()
```

```python
from pyspark.ml.feature import Imputer

imputer = Imputer(
    inputCols=['AirQuality', 'WaterPollution'], 
    outputCols=["{}_imputed".format(c) for c in ['AirQuality', 'WaterPollution']]
    ).setStrategy("mean")
```

```python
# add imputation cols to df 
imputer.fit(df_pyspark).transform(df_pyspark).show()
```
