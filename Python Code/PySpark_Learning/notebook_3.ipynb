PySpark Dataframes: filter operation, &,|,== and ~.

```python
from pyspark.sql import SparkSession

import os
os.environ["JAVA_HOME"] = r"C:\Program Files\Eclipse Adoptium\jdk-17.0.19.10-hotspot"
os.environ["PATH"] = os.environ["JAVA_HOME"] + r"\bin;" + os.environ["PATH"]
```

```python
#Data operation
spark = SparkSession.builder.appName('Practise').getOrCreate()
```

```python
df_pyspark = spark.read.csv('Water Quality - Every Drop Matters\Cities3.csv',header=True,inferSchema=True)
df_pyspark.show()
```

```python
#Filter operations
df_pyspark.filter("AirQuality>=50").show()
```

```python
df_pyspark.filter("AirQuality>=50").select(['City','Region']).show()
```

```python
df_pyspark.filter(df_pyspark['AirQuality']>=50).show()
```

```python
#How to use two different conditions 
df_pyspark.filter(
    (df_pyspark['AirQuality']>=50) 
    & (df_pyspark['WaterPollution']>=10)
).show()
```

```python
#Not operation
df_pyspark.filter(
    ~(df_pyspark['AirQuality']>=50) 
).show()
```
