PySpark GroupBy and aggregate function

```python
from pyspark.sql import SparkSession

import os
os.environ["JAVA_HOME"] = r"C:\Program Files\Eclipse Adoptium\jdk-17.0.19.10-hotspot"
os.environ["PATH"] = os.environ["JAVA_HOME"] + r"\bin;" + os.environ["PATH"]
```

```python
spark = SparkSession.builder.appName('Practise').getOrCreate()
```

```python
df_pyspark = spark.read.csv('Water Quality - Every Drop Matters\Cities3.csv',header=True,inferSchema=True)
df_pyspark.show()
```

```python
df_pyspark.printSchema()
```

```python
#Group by operation 
df_pyspark.groupBy('City').sum()
```

```python
df_pyspark.groupBy('City').sum().show()
```

```python
df_pyspark.groupBy('Country').sum().show()
```

```python
df_pyspark.groupBy('Country').mean().show()
```

```python
df_pyspark.groupBy('Country').count().show()
```

```python
df_pyspark.agg({'AirQuality':'sum'}).show()
```

```python
df_pyspark.groupBy('Country').max().show()
```

```python
df_pyspark.groupBy('Country').min().show()
```
